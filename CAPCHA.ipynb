```python
import random
import string

def generate_captcha(length=5):
    # Combine uppercase, lowercase, and digits
    characters = string.ascii_letters + string.digits
    # Generate a random combination of the specified length
    captcha = ''.join(random.choices(characters, k=length))
    return captcha

# Generate and display the CAPTCHA
print("Generated CAPTCHA:", generate_captcha())
```
